```python
import matplotlib.pyplot as plt
import cv2
import matplotlib.image as im
import numpy as np
```

```python
img = im.imread('Q10.png')
```

```python
plt.imshow(img,cmap='gray')
plt.show()
```

```python

img[(img[:,:,0] > 0.6) | (img[:,:,1] > 0.6) | (img[:,:,2] > .6)] =1
```

```python
plt.imshow(img,cmap='gray')
```

```python
#Erosion
kernel = np.ones((2,2),np.uint8)
img_erosion = cv2.erode(img,kernel,iterations=1)
plt.imshow(img_erosion)
```

```python
imgg = cv2.cvtColor(cv2.cvtColor(img,cv2.COLOR_RGB2BGR),cv2.COLOR_BGR2GRAY)
```

```python
plt.imshow(imgg)
```

```python

plt.imsave('Q10.png',img_erosion)
```
