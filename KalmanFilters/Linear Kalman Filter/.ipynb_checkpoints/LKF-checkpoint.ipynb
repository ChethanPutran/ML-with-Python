##                        2D Tracking System Model

State Vector, $x_k = \begin{bmatrix}p_x\\p_y\\v_x\\v_y\end{bmatrix}$

Noise Vector, $w_k = \begin{bmatrix}a_x\\a_y\end{bmatrix} a_x,a_y => N(0,\sigma^2) $


### 1. Discrete Process Modal (Prediction Step)

$x^-_k = Fx^+_{k-1} +  G_{k-1}u_{k-1} + Lw_{k-1}$

$x^-_k$ - State prediction\
$F$ - State Transition Matrix \
$L$  - Noise Input Matrix\
$G$ - Control Input Matrix

Let no control input is applied, then $G = 0$


$ \begin{bmatrix}p_x\\p_y\\v_x\\v_y\end{bmatrix}_k =  \begin{bmatrix} 1 & 0 & \delta{t} & 0 \\0 & 1 &  0 & \delta{t} \\ 0 & 0 & 1 & 0 \\ 0 & 0 & 0 & 1 \end{bmatrix} \begin{bmatrix} p_ x \\ p_y \\ v_x \\ v_y \end{bmatrix}_{k-1} +  \begin{bmatrix}\frac{1}{2}\delta{t^2} & 0 \\ 0 & \frac{1}{2} \delta{t^2} \\ \delta{t} & 0 \\  0 & \delta{t}  \end{bmatrix}  \begin{bmatrix}a_x\\a_y\end{bmatrix}_{k-1}$


### Covariance Propagation:


To account for new uncertainity in hte system due to the process model noise we add a new covariance term:\
\
$P_k^- = F_{k-1}P_{k-1}^+F_{k-1}^T + L_{k-1}Q_{k-1}L_{k-1}^T$


For simplicity,
$\sigma_x = \sigma_y = \sigma$, then
$x_k = Fx_{k-1} +  G_{k-1}u_{k-1} + \hat{w}_{k-1}$

Where,

$\hat{w}_{k-1} = L{w}_{k-1}$ -  Additive noise

$Q_k = \sigma \begin{bmatrix} 1 & 0 \\ 0 & 1 \end{bmatrix}$ -> Increases the uncertainity

=> $P_k^- = F_{k-1}P_{k-1}^+F_{k-1}^T + \hat{Q}$

<br />

Where,

$P_k$- Process covariance,\
$P_k^-$ - Predicted covariance\
$P_k^+$ - Previous uncertainity\
$\sigma$ - Measurement standard deviation\
$Q_k$ - Noise covariance(Covariance of process model) =$\begin{bmatrix}\sigma^2_x & 0 \\ 0 & \sigma^2_y \end{bmatrix}$



$\hat{Q} = \sigma^2 \begin{bmatrix}\frac{1}{2}\delta{t^2} & 0 & 0 & 0 \\ 0 & \frac{1}{2} \delta{t^2} & 0 & 0 \\0 & 0 & \delta{t} & 0 \\  0 & 0 & 0 &\delta{t}  \end{bmatrix} $



If we know exact value of $x_0$ then $P^+_0 = 0$ (there is no uncertainity)\
If no knowledge of $x_0$ then $P^+_0 -> \infty $ (there is large uncertainity)



Approximate value of $x_0$ we set as,
    $P^+_0 = E[ (x_0 - \hat{x}^+_0) (x_0 - \hat{x}^+_0)^T ]$
    
### Checking 


#### * When velocity uncertainity is set to non-zero and positional uncertainity is zet to zero, then positional error increases graduallly 
####  * When the position uncertainity is set to non-zero and velocity uncertainity is zet to zero, then velocity error increases gradually

#### * When position and velocity uncertainity in set to zero and the acceleration std or process-model noise to non-zero, then total level of uncertainity grows gradually
#### (velocity->linearly(integ of acc),position-quadratically(double integ of acc)

### 2. Updation (Measurement Model) 

$z_k = H_kx_k + M_kv_k$

Where,

$z_k$ - Current measurement\
$x_k$ - State variable\
$v_k$ - Additive noise\
$H_k$ - Measurement model matrix = $\begin{bmatrix} 1 & 0 & 0 & 0 \\ 0 & 1 & 0 & 0\end{bmatrix}$\
$M_k$ - Measurement noise sensitivity matrix

$\hat{x}^-_k$ - State prediction\
$\hat{x}^+_k$ - Updated state estimate

Updation is done based on the error between the measurement $z_k$ and predicted measurement calculated from $H_k\hat{x}^-_k$

### Innovation :
$\tilde{y}_k = z_k - H_k\hat{x}^-_k$\
$\tilde{y}_k = z_k - \tilde{z}_k$


### Innovation Covariance :
$S_k = H_kP^-_kH^T_k + R_k$


Since $v_k$ is additive,

$R_k = M_kR_kM^T_k = \sigma^2 \begin{bmatrix} 1 & 0 \\ 0 & 1 \end{bmatrix}$

Where,

$\sigma$ - Measurement Std.\
$\tilde{z}_k$ - Predicted measurement Vector\
$\tilde{y}_k$ - Measurement innovation vector\
$P^-_k$ - Covariance of state(Uncertainity in state)\
$H_kP^-_kH^T_k$ - Covariance transformation from $x_k$ to $y_k$\
$R_k$ - Covariance of noise of measurement \
$S_k$ - Innovation covariance matrix


#### The state is updated based on the size of the innovation $\hat{y_k}$, using  the Kalman Gain matrix $K_k$, which itself is formed as a ratio between current state uncertainity and  the innovation uncertainity

### State Update :

$\hat{x}^+_k = \hat{x}^-_k + K_k\tilde{y}_k $

### Kalman Gain :
$ K_k = P^-_kH^T_kS^{-1}_k$


### Covariance Update :

$P^+_k = (I - K_kH_k)P^-_k $
